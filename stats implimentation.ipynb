```python
import numpy as np

def outlier_detection(data, threshold=2):
    mean_data = np.mean(data)
    std = np.std(data)

    outliers = [
        x for x in data
        if abs(x - mean_data) / std > threshold
    ]

    return outliers


# Example
data = [1, 2, 3, 4, 5, 7, 834, 56, 78]

print("Outliers:", outlier_detection(data))
```

```python
import numpy as np

def outlier_detection(data, threshold=2):
    mean_data = np.mean(data)
    std = np.std(data)

    outliers = [
        x for x in data
        if abs(x - mean_data) / std > threshold
    ]

    return outliers


# Example
data = [1, 2, 3, 4, 5, 7, 834, 56, 78]

print(outlier_detection(data))
```
